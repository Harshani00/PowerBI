```python
import pandas as pd
from scipy import stats
import numpy as np
```

```python
df = pd.read_csv('sales_data_sample.csv', encoding='cp1252')
```

```python
df.info()
```

```python
# Rename columns
df = df.rename(columns={
    'ORDERNUMBER': 'Order No',
    'QUANTITYORDERED': 'Qty Ordered',
    'PRICEEACH': 'Price Each',
    'ORDERLINENUMBER': 'Order Line No',
    'SALES': 'Sales',
    'ORDERDATE': 'Order Date',
    'STATUS': 'Status',
    'QTR_ID': 'Quarter',
    'MONTH_ID': 'Month',
    'YEAR_ID': 'Year',
    'PRODUCTLINE': 'Product Line',
    'MSRP': 'MSRP',
    'PRODUCTCODE': 'Product Code',
    'CUSTOMERNAME': 'Customer Name',
    'PHONE': 'Phone',
    'ADDRESSLINE1': 'Address 1',
    'ADDRESSLINE2': 'Address 2',
    'CITY': 'City',
    'STATE': 'State',
    'POSTALCODE': 'Postal Code',
    'COUNTRY': 'Country',
    'TERRITORY': 'Territory',
    'CONTACTLASTNAME': 'Contact Last Name',
    'CONTACTFIRSTNAME': 'Contact First Name',
    'DEALSIZE': 'Deal Size'
})
```

```python
df.info()
```

```python
z_scores_qty = np.abs(stats.zscore(df['Qty Ordered']))
outliers_qty = df[z_scores_qty > 3]
print(outliers_qty)
```
